# 4) filtro por região

## 5) criando o mapa interativo com folium

### o que faz o folium

- ele é uma biblioteca Python usada para criar mapas interativos, permitindo adicionar marcadores, pop-ups e camadas no mapa.


```python
# aqui acontece a criação do mapa centralizado no Brasil
mapa = folium.Map(
    location=[-14.2350, -51.9253],
    zoom_start=4
)

# adicionamos capitais como um marcador no mapa
for linha in df.itertuples():

    # aqui faremos o que vai aparecer quando clicar no marcador
    texto = f"""
    <b>{linha.capital}</b><br>
    Estado: {linha.estado}<br>
    UF: {linha.uf}<br>
    Região: {linha.regiao}<br>
    Latitude: {linha.latitude}<br>
    Longitude: {linha.longitude}
    """

    # fizemos o marcador no mapa
    folium.Marker(
        location=[linha.latitude, linha.longitude],
        popup=folium.Popup(texto, max_width=300),
        tooltip=linha.capital
    ).add_to(mapa)

# mostra a pre-visualização
mapa
```

# 6) adicionando o filtro

Aqui fizemos o mesmo codigo, no entanto adicionando os filtros

```python

mapa_regioes = folium.Map(
    location=[-14.2350, -51.9253],
    zoom_start=4
)

# aqui faremos uma camada para cada região
for regiao in df["regiao"].unique():

    # usamos o FeatureGroup para conseguir criar um grupo de marcadores
    grupo_regiao = folium.FeatureGroup(name=regiao)

    # filtramos as capitais daquela região
    dados_regiao = df[df["regiao"] == regiao]

    # colocamos as capitais de suas respectivas regiões no grupo indicado a elas
    for linha in dados_regiao.itertuples():

        texto_popup = f"""
        <b>{linha.capital}</b><br>
        Estado: {linha.estado}<br>
        UF: {linha.uf}<br>
        Região: {linha.regiao}<br>
        Latitude: {linha.latitude}<br>
        Longitude: {linha.longitude}
        """

        folium.Marker(
            location=[linha.latitude, linha.longitude],
            popup=texto_popup,
            tooltip=linha.capital
        ).add_to(grupo_regiao)

    # adicionamos o grupo no mapa
    grupo_regiao.add_to(mapa_regioes)

# aqui temos o controle para mostrar ou nao determinada região no mapa
folium.LayerControl(collapsed=False).add_to(mapa_regioes)


mapa_regioes
```
